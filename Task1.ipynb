```python
import pandas as pd

df = pd.read_excel('/Online Retail.xlsx')

print("Total rows:", len(df))
print("Total columns:", len(df.columns))

display(df.head())
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
before = len(df)

df = df.drop_duplicates()

after = len(df)

print("Rows before cleaning:", before)
print("Duplicate rows removed:", before - after)
print("Rows after cleaning:", after)
```

```python
print(df.dtypes)
```

```python
df["CustomerID"] = df["CustomerID"].astype("Int64")

print(df.dtypes)
```

```python
df["Description"] = df["Description"].fillna("Unknown product")

print("Missing descriptions:", df["Description"].isnull().sum())
print("Unknown product entries:", (df["Description"] == "Unknown product").sum())
```

```python
print("Zero quantity:", (df["Quantity"] == 0).sum())
print("Negative quantity:", (df["Quantity"] < 0).sum())
print("Zero or negative price:", (df["UnitPrice"] <= 0).sum())

display(df[["Quantity", "UnitPrice"]].describe())
```

```python
before = len(df)

df = df[df["UnitPrice"] > 0].copy()

print("Rows before price cleaning:", before)
print("Invalid-price rows removed:", before - len(df))
print("Rows after price cleaning:", len(df))
print("Remaining invalid prices:", (df["UnitPrice"] <= 0).sum())
```

```python
# Calculate transaction value
df["TransactionAmount"] = df["Quantity"] * df["UnitPrice"]

# Extract date features
df["Year"] = df["InvoiceDate"].dt.year
df["Month"] = df["InvoiceDate"].dt.month
df["DayOfWeek"] = df["InvoiceDate"].dt.day_name()

# Identify cancellations and returns
df["IsCancellation"] = df["InvoiceNo"].astype(str).str.startswith("C")
df["IsReturnQuantity"] = df["Quantity"] < 0

display(df.head())

print("Total rows:", len(df))
print("Total columns:", len(df.columns))
```

```python
print("Rows:", len(df))
print("Columns:", len(df.columns))

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nInvalid prices:", (df["UnitPrice"] <= 0).sum())
```

```python
df.to_csv("/content/cleaned_online_retail.csv", index=False)

print("Dataset saved successfully!")
print("File: cleaned_online_retail.csv")
```
